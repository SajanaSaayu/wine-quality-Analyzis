```python
#Load the dataset and handle any missing data by replacing null values with the mean value of the respective column. (Score 2)
import pandas as pd
import numpy as np

# Load the dataset
wine_data = pd.read_csv("winequality-red.csv", sep=";")

# Check missing values
print("Missing values before handling:")
print(wine_data.isnull().sum())

# Replace missing values with the mean of respective columns
wine_data.fillna(wine_data.mean(), inplace=True)

# Check missing values after handling
print("\nMissing values after handling:")
print(wine_data.isnull().sum())

# Display the dataset
wine_data.head()
```

```python
#Extract the following columns as vectors: alcohol,citric acid. 

# Extract alcohol column as a vector
alcohol = wine_data["alcohol"].values

# Extract citric acid column as a vector
citric_acid = wine_data["citric acid"].values

# Display the vectors
print("Alcohol vector:")
print(alcohol)

print("\nCitric Acid vector:")
print(citric_acid)
```

```python
#Select two features (e.g., alcohol and density) from the dataset and calculate the covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the selected columns.(Score 2)

# Select two features
X = wine_data[['alcohol', 'density']].values

# Calculate covariance matrix
cov_matrix = np.cov(X.T)

# Display the covariance matrix
print("Covariance Matrix:")
print(cov_matrix)
```

```python
#Perform eigen decomposition on the covariance matrix you computed in question 3.
#Identify and interpret the results:Identify the top 2 eigenvalues of the covariance matrix,Identify the corresponding eigenvectors.(Score 2)

# Perform eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Display eigenvalues
print("Eigenvalues:")
print(eigenvalues)

# Display eigenvectors
print("\nEigenvectors:")
print(eigenvectors)

# Sort eigenvalues in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]

# Arrange eigenvalues from largest to smallest
top_eigenvalues = eigenvalues[sorted_indices]

# Arrange corresponding eigenvectors
top_eigenvectors = eigenvectors[:, sorted_indices]

# Display top 2 eigenvalues
print("\nTop 2 Eigenvalues:")
print(top_eigenvalues)

# Display corresponding eigenvectors
print("\nCorresponding Eigenvectors:")
print(top_eigenvectors)
```

```python
# Count each quality score
quality_counts = wine_data["quality"].value_counts()

# Find the quality score with the highest frequency
highest_frequency_quality = quality_counts.idxmax()

print("Most common wine quality:", highest_frequency_quality)
```
